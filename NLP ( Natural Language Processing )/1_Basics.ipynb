# NLP Basics: Making Machines Understand Human Language

NLP stands for Natural Language Processing. It is the field of AI that helps computers understand, process, and generate human language.

Think of NLP as the translator between two very different worlds:
- Humans speak in words, sentences, emotions, sarcasm, and random spelling mistakes.
- Computers understand numbers, not feelings. Shocking, but true.

So NLP is basically the process of turning messy human language into something a machine can work with.

![NLP overview](images/basics_nlp.svg)

## Why do we need NLP?

Humans communicate using language, but computers only understand structured data. NLP bridges that gap.

Examples:
- Google search understands your query
- Alexa or Siri recognize your voice
- Chatbots respond intelligently
- Spam filters detect suspicious emails
- Sentiment analysis reads customer opinions

## Core idea

A machine cannot read words the way a human does. It needs text converted into a form it can process mathematically.

That is why NLP is so important in real-world AI systems.

## Main approaches in NLP

1. Rule-based approach
   - Uses hand-written rules
   - Good for simple tasks
   - Not very flexible for complex language

2. Machine learning approach
   - Learns patterns from data
   - Used in text classification and sentiment analysis
   - Needs feature extraction before training

3. Deep learning approach
   - Uses neural networks for better context understanding
   - Powerful for chatbots, translation, and language models
   - Often uses embeddings and transformers

## Important note

Language is messy. Humans say things like:
- "That was great!"
- "Yeah, sure..."
- "I am not unhappy"

A machine needs structure to understand these patterns. That is exactly what NLP tries to do.

> In short: NLP helps machines understand human language without making them read a dictionary and cry.

# Approaches of NLP

## 1. Rule-Based Approach (Old School NLP)

"Early method of solving NLP using handwritten rules and grammer logic."

- You write rules manually to process and understand language.

- These rules can define grammer, spellings, corrections, sentence structures, etc.

### Example:
A Rule like:

"If sentence contains 'not good', mark it as negative sentiment."

### Used for:
- Grammer checkers
- Spell check
- Simple keyword-based chatbots

### Limitations
- Can't handle complex, ambigious, or new patterns
- Not scalable for large data

## 2. Statistical ( Machine Learning ) Approach:
"Language is treated like data, and algorithms learn patterns from it."
- Use of models like 
    - Naive Bayes
    - Logistic Regression
    - SVM ( Support Vector Machines )

Requires text to be converted into numbers using techniques like Bag of Words, TF-IDF, etc.
### Example:
Train a model on 10,000 product reviews. It learns which words indicates positive or negative sentiment and predicts for new reviews.

### Used for:
- Sentiment Analysis
- Spam Detection
- Text Classification

### Limitations: 
- Loses context of words ( like sarcasm or sequence of words )

## 3. Deep Learning-Based Approach ( Modern NLP ):
"Use neural networks to automatically learn complex patterns and context from text."
- models can understand meaning, context, word order, and more.
- Popular Architectures:
    - RNN ( Recurrent Neural Networks )
    - LSTM ( Long Short-Term Memory )
    - GRU ( Gated Recurrent Unit )
    - Transformers ( BERT, GPT, T5, etc. )

Even in deep learning, we have to convert text into numbers but we don't use traditional techniques like bag of words or TF-IDF. Instead, we use more advanced methods like Word Embeddings ( Word2Vec, GloVe ) or Token IDs with Embedding Layers, which actually captures the meaning and context of the words.
