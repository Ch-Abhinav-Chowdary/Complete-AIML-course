```python
# This reduces variations such as 'Happy' and 'happy'
df['text'] = df['text'].apply(lambda x: x.lower())
df
```

```python
import string

def remove_punc(txt):
    return txt.translate(str.maketrans('', '', string.punctuation))
```

```python
# Apply punctuation removal to every text entry
df['text'] = df['text'].apply(remove_punc)
```

```python
df
```

```python
# Define a function to remove numeric characters from the text
# This helps keep the text focused on words rather than digits
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

# Apply the function to all rows

df['text'] = df['text'].apply(remove_numbers)
```

```python
# Define a helper function to remove emojis and non-ASCII characters
# This keeps the text cleaner and more usable for NLP models
def remove_emoji(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

# Apply the cleaning function to the text column
df['text'] = df['text'].apply(remove_emoji)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
stop_words
```

```python
df.loc[1]['text']
```

```python
# Removing the stopwords

def remove(txt):
    words = txt.split()
    cleaned = []
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
    
    return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
df['text'], df['label'], test_size=0.2, random_state=42, 
stratify=df['label'])
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
Xtr = cv.fit_transform(X_train)
Xte = cv.transform(X_test)

Xtr.shape
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(Xtr, y_train)
```

```python
y_pred = model.predict(Xte)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

classification_report(y_test, y_pred)
```

```python
accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
import numpy as np
words = np.array(cv.get_feature_names_out())
spam_idx = list(model.classes_).index('spam')
scores = model.feature_log_prob_[spam_idx] - model.feature_log_prob_[1-spam_idx]
print('Top spam words:', words[np.argsort(scores)[-15:]])
print('Top ham words :', words[np.argsort(scores)[:15]])
```

```python
import joblib
from sklearn.pipeline import Pipeline

model_pipe = Pipeline([('cv', CountVectorizer()), ('nb', MultinomialNB())])
model_pipe.fit(X_train, y_train)

joblib.dump(model, 'spam_model.pkl')

def check(msg):
    return model_pipe.predict([msg])[0]

print(check('congratulations You won a free ticket call this number now'))     # Make sure to send a clean data
print(check('are we meeting at the library at five'))
```
