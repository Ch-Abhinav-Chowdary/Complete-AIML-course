# Feature Extraction / Vectorization

Feature extraction is the process of converting text into numerical features so that a machine learning model can understand it.

A model does not read words like a human. It only sees numbers. So text must be transformed before learning begins.

![Feature extraction](images/feature_extraction.svg)

## Why feature extraction is needed

Language is symbolic. Words are not directly useful to an algorithm unless we represent them in a machine-readable form.

Example:
- raw text: "I love NLP"
- after preprocessing: ["I", "love", "NLP"]
- after vectorization: numbers representing the text

## Common techniques

### 1. One-hot encoding
- Represents each word as a binary vector
- Simple but high-dimensional and inefficient

### 2. Bag of Words
- Counts the frequency of words
- Ignores word order and context
- Easy to understand but not very sophisticated

### 3. TF-IDF
- Measures how important a word is in a document
- Better than simple frequency counting
- Commonly used in text classification

### 4. Word2Vec
- Learns word meanings from context
- Captures similarity between words
- More powerful than traditional methods

### 5. Transformers / BERT
- Understand context and sentence meaning better
- Used in modern NLP systems
- Powerful but heavier computationally

## Key takeaway

Feature extraction is the bridge between human language and machine learning.

Without it, your model would be staring at words like a confused robot and pretending to understand them.

> In simple words: if text is the story, vectorization is the language the machine actually speaks.

# Feature Extraction / Vectorization — Methods, Examples, Pros, and Cons

We already know text must become numbers. Now let’s explore the most common ways to do that, with real examples and honest pros/cons.

### 1. One-Hot Encoding

Definition:
- Each unique word gets its own position in a vector.
- The word is represented by 1 at its position and 0 everywhere else.

Example vocabulary: ["I", "love", "NLP", "you"]
- "I love NLP" → [1, 1, 1, 0]
- "I love you" → [1, 1, 0, 1]

Pros:
- Very simple to understand and implement
- Works well for very small vocabularies
- No assumptions about word similarity

Cons:
- Vector size equals vocabulary size, which explodes quickly
- No meaning or similarity captured between words
- Every sentence is sparse and mostly zeros

### 2. Bag of Words (BoW)

Definition:
- Count each word in a document.
- Order of words is ignored.

Example with vocabulary ["I", "love", "NLP", "you"]:
- "I love NLP" → [1, 1, 1, 0]
- "I love you" → [1, 1, 0, 1]
- "I love NLP NLP" → [1, 1, 2, 0]

Pros:
- Easy and fast to compute
- Useful for simple text classification tasks
- Good for small to medium sized datasets

Cons:
- Ignores word order and context
- Treats all words as independent features
- High dimensional vectors and sparse data

### 3. TF-IDF (Term Frequency - Inverse Document Frequency)

Definition:
- Combines word frequency with how rare a word is across documents.
- Common words like "the" are downweighted.

Example idea:
- Term Frequency (TF): how often a word appears in the document.
- Inverse Document Frequency (IDF): log(total documents / documents containing the word).
- TF-IDF = TF * IDF.

Example:
- Document A: "I love NLP"
- Document B: "I love machine learning"
- "love" appears in both documents, so it gets lower weight.
- "NLP" appears only in A, so it gets higher weight.

Pros:
- Reduces importance of common words
- Better for keyword-based classification
- Short texts still work better than raw counts

Cons:
- Still ignores word order
- Cannot capture semantic meaning or word similarity
- Sensitive to document size and noise

### 4. Word2Vec

Definition:
- Learns dense word vectors from text.
- Words that appear in similar contexts have similar vectors.

Example:
- "king" and "queen" may end up close in vector space.
- "cat" and "dog" are likely closer than "cat" and "blue".

Pros:
- Captures similarity between words
- Produces compact dense vectors
- Good for semantic meaning and analogy tasks

Cons:
- Requires a large corpus to learn useful embeddings
- Still ignores sentence structure beyond local context
- More complex than simple count methods

### 5. Transformers / BERT

Definition:
- Uses contextual embeddings based on the whole sentence.
- Each word gets a vector that depends on its surrounding words.

Example:
- "I love NLP" and "I love my dog" both know "love" means affection, but BERT learns context.
- The vector for "bank" in "river bank" differs from "bank account".

Pros:
- Captures context and word meaning in full sentences
- State-of-the-art performance for many tasks
- Works well for sentence-level classification and QA

Cons:
- Computationally expensive and slow
- Large models require more memory and resources
- Overkill for very small or simple tasks

## Quick comparison

- One-Hot: simplest, huge vectors, no meaning
- BoW: counts words, ignores order
- TF-IDF: counts + rarity, better for keywords
- Word2Vec: semantic similarity, dense vectors
- BERT: full context, powerful but heavy

> If you want speed and simplicity, choose BoW/TF-IDF. If you want meaning and context, choose Word2Vec or BERT. Unless you just want to impress people with a Transformer name in your notebook.
