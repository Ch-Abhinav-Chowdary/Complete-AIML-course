```python
from collections import Counter

sentences = [
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

# Build vocabulary from all unique words
vocab = sorted({word.lower() for sentence in sentences for word in sentence.split()})
print("Vocabulary:", vocab)

# Create bag-of-words vectors manually
bow_vectors = []
for sentence in sentences:
    counts = Counter(word.lower() for word in sentence.split())
    vector = [counts.get(word, 0) for word in vocab]
    bow_vectors.append((sentence, vector))

for sentence, vector in bow_vectors:
    print(sentence, "->", vector)
```
