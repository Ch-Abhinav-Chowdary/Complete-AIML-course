```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample documents for TF-IDF
texts = [
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

# Create TF-IDF vectors
vectorizer = TfidfVectorizer()
X_tfidf = vectorizer.fit_transform(texts)

print("Vocabulary:", vectorizer.get_feature_names_out())
print("\nTF-IDF Matrix:\n", X_tfidf.toarray())
```

# TF-IDF

TF-IDF stands for Term Frequency-Inverse Document Frequency. It gives more importance to words that appear often in a document but are not common across many documents.

- Term Frequency: $TF(t, d) = \frac{\text{count of term } t \text{ in document } d}{\text{total terms in } d}$
- Inverse Document Frequency: $IDF(t, D) = \log\left(\frac{N}{DF(t)}\right)$
- TF-IDF score: $TF\text{-}IDF(t, d, D) = TF(t, d) \times IDF(t, D)$

## Example problem

Consider these documents:
1. I love pizza
2. Pizza is the best
3. I love pasta
4. Pasta is great

Here, words like "pizza" and "pasta" are meaningful for their own documents, while common words like "is" and "the" get lower weight.

## Pros and cons

Pros:
- Gives more weight to important words
- Helps reduce the impact of common words
- Works well for search and text similarity

Cons:
- Can still miss word order and context
- Depends heavily on document size and vocabulary
- May not capture semantics as well as embeddings

```python
from math import log

texts = [
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

# Build vocabulary
vocab = sorted({word.lower() for text in texts for word in text.split()})
print("Vocabulary:", vocab)

# Compute document frequency for each term
N = len(texts)
df = {term: sum(1 for text in texts if term in text.lower().split()) for term in vocab}

# Compute TF-IDF manually
for i, text in enumerate(texts):
    words = text.lower().split()
    total_terms = len(words)
    tfidf_vector = []
    for term in vocab:
        tf = words.count(term) / total_terms
        idf = log(N / df[term])
        tfidf_vector.append(round(tf * idf, 3))
    print(f"{text} -> {tfidf_vector}")
```
