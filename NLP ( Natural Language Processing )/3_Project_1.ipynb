# Project 1: Emotion Classification from Text

This project is a classic NLP task where we try to classify text into different emotion categories.

The idea is simple:
1. Load the dataset
2. Clean the text
3. Convert emotions into numeric labels
4. Train a model to predict the emotion of new text

![Emotion classification project](images/project1_emotion.svg)

## Project workflow

### 1. Load the dataset
We read the text file where each row contains a sentence and its emotion label.

### 2. Inspect the data
We check for missing values and understand the distribution of emotions.

### 3. Convert labels
Machine learning models cannot learn directly from words like "happy" or "sad". So we convert them into numeric values.

### 4. Text preprocessing
We perform steps like:
- lowercase conversion
- punctuation removal
- number removal
- emoji removal
- stopword removal

### 5. Model training
After preprocessing, the text becomes more structured and is ready for a classifier.

## Why this matters

Emotion detection is useful in:
- customer feedback analysis
- social media monitoring
- chatbots
- mental health applications

## Important note

A model does not feel emotions. It only learns patterns in words that often appear with certain labels.

So if the text says "I am so sad", the model learns the pattern and predicts sadness. It is not being emotional. It is just being very good at math and very bad at life.

> The project teaches one important lesson: clean text + good preprocessing + proper labels = a model that actually works.

```python
# Import libraries for data handling, visualization, and NLP tasks
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset from the text file using ';' as the separator
# The file contains two columns: text and emotion label
df = pd.read_csv('train.txt', sep=';', header=None, names=('text', 'emotion'))
```

```python
df.head()
```

```python
# Check whether the dataset contains any missing values
# This helps ensure the data is clean before training a model
df.isnull().sum()
```

```python
# Convert emotion labels into numeric values so the model can train on them
# Each unique emotion is assigned a unique integer ID
unique_emotions = df['emotion'].unique()

emotion_numbers = {}

i = 0

for emo in unique_emotions:
    emotion_numbers[emo] = i
    i += 1

# Map the original emotion names to numeric labels
df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
df
```

```python
# Convert all text to lowercase to make the data more consistent
# This reduces variations such as 'Happy' and 'happy'
df['text'] = df['text'].apply(lambda x: x.lower())
```

```python
df
```

```python
# Define a helper function to remove punctuation from the text
import string


def remove_punc(txt):
    return txt.translate(str.maketrans('', '', string.punctuation))
```

```python
# Apply punctuation removal to every text entry
df['text'] = df['text'].apply(remove_punc)
```

```python
# Define a function to remove numeric characters from the text
# This helps keep the text focused on words rather than digits
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

# Apply the function to all rows

df['text'] = df['text'].apply(remove_numbers)
```

```python
# Define a helper function to remove emojis and non-ASCII characters
# This keeps the text cleaner and more usable for NLP models
def remove_emoji(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

# Apply the cleaning function to the text column
df['text'] = df['text'].apply(remove_emoji)
```

```python
# What is NLTK

import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Removing Stopwords
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
stop_words
```

```python
df.loc[1]['text']
```

```python
# Removing the stopwords

def remove(txt):
    words = txt.split()
    cleaned = []
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
    
    return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# 1. Initialize the vectorizer
bow_vectorizer = CountVectorizer()

# 2. Fit and transform the training data
X_train_bow = bow_vectorizer.fit_transform(X_train)

# 3. ONLY transform the test data (Do not fit again!)
X_test_bow = bow_vectorizer.transform(X_test) 

# 4. Train the model
nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)

# 5. Predict
pred_bow = nb_model.predict(X_test_bow)
print(accuracy_score(y_test, pred_bow))
# 6. Evaluate (Optional, but highly recommended since you imported accuracy_score)
# print("Accuracy:", accuracy_score(y_test, pred_bow))
```

```python
tfidf_vectorizer = TfidfVectorizer()

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)
```

```python
nb_model2 = MultinomialNB()

nb_model2.fit(X_train_tfidf, y_train)
```

```python
pred_tfidf = nb_model2.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test, pred_tfidf))
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(max_iter=1000)

logistic_model.fit(X_train_tfidf, y_train)

y_pred = logistic_model.predict(X_test_tfidf)

print(accuracy_score(y_test, y_pred))
```
