```python
# One-Hot Encoding

# One-hot encoding converts each word into a binary vector based on a vocabulary. In this example, each word is represented by a vector with 1 in its position and 0 elsewhere.

import numpy as np

words = ["apple", "banana", "apple", "orange", "banana", "grape"]

# Build vocabulary
vocab = sorted(set(words))
print("Vocabulary:", vocab)

# Create mapping from word to index
word_to_index = {word: i for i, word in enumerate(vocab)}

# One-hot encode each word
encoded_words = []
for word in words:
    vector = np.zeros(len(vocab), dtype=int)
    vector[word_to_index[word]] = 1
    encoded_words.append((word, vector))

# Display result
for word, vector in encoded_words:
    print(word, "->", vector)
```
