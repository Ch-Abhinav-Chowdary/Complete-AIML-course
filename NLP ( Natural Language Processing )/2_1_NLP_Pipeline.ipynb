# NLP Pipeline: From Raw Text to Useful Intelligence

An NLP pipeline is the step-by-step process of taking raw text and turning it into something a model can learn from.

If you skip preprocessing, your model will learn nonsense dressed in fancy clothes. That is not intelligence. That is just garbage with a nice haircut.

![NLP pipeline](images/nlp_pipeline.svg)

## Step 1: Data collection

We first gather text data from:
- public datasets
- web scraping
- APIs
- manual collection

## Step 2: Text cleaning

Clean text before feeding it to a model.

Common cleaning operations:
- lowercase all text
- remove punctuation
- remove numbers if unnecessary
- remove URLs and HTML tags
- remove emojis and special characters
- remove stopwords
- correct spelling if needed

## Step 3: Tokenization

Tokenization means splitting text into smaller units such as words or sentences.

Example:
- "I love NLP" → ["I", "love", "NLP"]

## Step 4: Vectorization

Machines do not understand words directly. So we convert text into numbers.

Popular methods include:
- Bag of Words
- TF-IDF
- Word2Vec
- BERT embeddings

## Step 5: Model training

Once text is converted into features, we train a machine learning or deep learning model for:
- classification
- sentiment analysis
- spam detection
- chatbot responses

## Golden rule

Good preprocessing leads to better results. Bad preprocessing leads to a model that confidently predicts the wrong thing.

> Remember: a model is only as smart as the data you give it. If you feed it messy text, it will behave like a confused intern.

#### Step-2: Text Cleaning
Now there are many things we do in Text Cleaning and depending on the task, Let's see some examples.

a) LowerCasing: Convert all text to lowercase.

b) Remove punctuations: Remove symbols like .,?!@#$%^&*()
- They don't add meaning in ML context.
- In deep learning-based NLP ( Especially with models like BERT, GPT, T5 ), punctuations can carry meaning and we usually keep it during preprocessing.

c) Removing Numbers: Remove digits from the text ( optional - depends on task )
- "I bought 5 phones" -> you may or may not need the number.

d) Remove URls/Links: Remove any links in the text ( very common in social media data )

e) Romove HTML Tags: If scraping from web, remove things like < div >, < p >, etc.

f) Remove Emojis and Special Characters: Emojis might break vectorization if not removed or handled

g) Remove Stopwords: Remove common words like "is", "the", "was", "and", "in", etc. These don't help in understanding the meaning.

h) ( Optional ) Spelling Corrections: Fix spelling mistakes using textBlob or other libraries.
