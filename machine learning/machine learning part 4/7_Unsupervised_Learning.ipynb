# Unsupervised Learning

## What is unsupervised learning?
Unsupervised learning is a branch of machine learning where the model is given data without labels and is asked to discover patterns, structure, or hidden relationships on its own.

Unlike supervised learning, there is no correct answer provided during training. The model must learn by exploring the data.

## Why it is called "unsupervised"
The word "unsupervised" means there is no teacher telling the model what the correct answer is. The model works like a student who looks at many examples and tries to find useful patterns by itself.

## Main goals of unsupervised learning
- Find hidden groups or clusters in the data
- Reduce the number of features while keeping important information
- Detect unusual or abnormal data points
- Understand the structure of the data
- Discover relationships between variables

## Common techniques in unsupervised learning

### 1) Clustering
Clustering means grouping similar data points together.
- Example: grouping customers with similar shopping behavior
- Example: grouping similar documents or images

### 2) Dimensionality Reduction
This technique reduces the number of features while keeping the most important information.
- Example: PCA (Principal Component Analysis)
- Useful when data has many columns and we want simpler representation

### 3) Anomaly Detection
Anomaly detection finds unusual points that do not match the normal pattern.
- Example: fraud detection in banking
- Example: detecting faulty machines in manufacturing

## Why unsupervised learning is useful
It helps us explore data when labels are not available.
This is very common in real-world situations because labeling data can be expensive, slow, or impossible.

## Real-world use cases
- Customer segmentation for marketing
- Grouping similar documents or news articles
- Recommender systems
- Fraud detection
- Image compression and feature reduction
- Medical data analysis

## Difference from supervised learning
- Supervised learning: learns from labeled data
- Unsupervised learning: discovers patterns from unlabeled data
