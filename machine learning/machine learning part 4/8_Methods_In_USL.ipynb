# Clustering Methods

## 1) K-Means Clustering
K-Means is a centroid-based clustering algorithm. It groups data into $k$ clusters by finding cluster centers, called centroids.

### How it works
1. Choose the number of clusters $k$.
2. Initialize $k$ centroids randomly.
3. Assign each data point to the nearest centroid.
4. Recalculate the centroid of each cluster.
5. Repeat until the centroids no longer change much.

### Main idea
Each point belongs to the cluster whose center is closest to it.

### Advantages
- Simple and fast
- Easy to understand
- Works well for large datasets

### Disadvantages
- You must choose $k$ manually
- Sensitive to initial centroid placement
- Works best when clusters are roughly spherical
- Struggles with irregular shapes

### Elbow Method
The elbow method helps choose the best value of $k$.
- Plot the within-cluster sum of squares (WCSS) against different values of $k$.
- The best $k$ is usually found at the point where the curve bends, called the elbow.

## 2) DBSCAN Clustering
DBSCAN is a density-based clustering algorithm. It groups points that are close to each other and labels sparse areas as noise.

### Key concepts
- Core point: a point that has enough neighboring points within a radius $\text{eps}$.
- Border point: a point near a cluster but not dense enough to be a core point.
- Noise: points that do not belong to any dense cluster.

### Important parameters
- $\text{eps}$: the radius around a point
- minPoints: the minimum number of points required to form a dense region

### Why DBSCAN is useful
- It can find clusters of arbitrary shapes.
- It can detect outliers naturally.
- It does not require the number of clusters to be specified in advance.

### Limitations
- Sensitive to the choice of $\text{eps}$ and minPoints.
- Can struggle when clusters have very different densities.

## 3) K-Means vs DBSCAN
- K-Means needs the number of clusters beforehand.
- DBSCAN does not need the number of clusters beforehand.
- K-Means is centroid-based.
- DBSCAN is density-based.
- K-Means is better for compact clusters.
- DBSCAN is better for irregular clusters and noise detection.

## 4) Study summary
Clustering is used to discover natural groups in data. K-Means is simple and efficient, while DBSCAN is better at handling irregular shapes and outliers.
