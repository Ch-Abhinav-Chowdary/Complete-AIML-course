# Cross Validation and Hyperparameter Tuning

## 1) What is validation in machine learning?
Validation is the process of checking whether a model generalizes well to new, unseen data. A model may perform very well on training data, but that does not guarantee good performance in real-world applications.

### Why validation matters
- Training accuracy can be misleading.
- A model may memorize the training examples instead of learning patterns.
- We need a reliable way to estimate future performance.

## 2) Cross Validation
Cross validation is a technique where the data is split into multiple parts, and the model is trained and tested several times.

### Main idea
- Split the data into $k$ folds.
- Train on $k-1$ folds.
- Validate on the remaining fold.
- Repeat this process until every fold has been used as validation data.
- Take the average score.

### Why it is useful
- Gives a more reliable estimate of performance.
- Reduces the chance of overfitting.
- Helps compare models more fairly.

### Common types
- Train-Test Split
- K-Fold Cross Validation
- Stratified K-Fold Cross Validation

### Example: 5-Fold Cross Validation
If you have 100 samples:
- Split the data into 5 equal parts.
- Use 4 parts for training and 1 part for testing.
- Repeat the process 5 times.
- Average the results.

This is much better than relying on one single train-test split.

## 3) Overfitting and Underfitting
- Overfitting: the model learns too much from the training data and fails on new data.
- Underfitting: the model is too simple and does not learn the patterns properly.

Cross validation helps us detect and reduce these problems.

## 4) Model Tuning
Model tuning means adjusting the settings of a model so it performs better.

Examples of model settings:
- Number of neighbors in KNN
- C and gamma in SVM
- Maximum depth of a decision tree

## 5) Hyperparameters
Hyperparameters are configuration values chosen before training begins.

They are different from model parameters because:
- Hyperparameters are set by the practitioner.
- Model parameters are learned from the data.

### Examples
- KNN: $k$ (number of neighbors)
- SVM: $C$, kernel, gamma
- Random Forest: number of trees, max depth

## 6) Hyperparameter Tuning Techniques

### A) Manual Search
Try values manually and compare results.
- Simple
- Slow
- Depends on experience

### B) Grid Search CV
Try all combinations from a predefined list.
- Very systematic
- Can be time-consuming

### C) Randomized Search CV
Try random combinations from the search space.
- Faster than grid search
- Often gives very good results

## 7) Practical workflow
1. Split the data.
2. Choose a model.
3. Define a search space for hyperparameters.
4. Use cross validation to evaluate the combinations.
5. Select the best parameters.
6. Retrain the final model on the full training data.

## 8) Simple summary
- Validation checks whether the model will work on new data.
- Cross validation gives a better estimate of performance.
- Hyperparameter tuning helps us choose the best settings.
- Better tuning usually leads to better generalization.
